## 4. NumPy practice

```python
# TODO: Select numeric columns relevant to your project
# TODO: Convert them to a NumPy array using .to_numpy()
# TODO: Calculate average, maximum and minimum
subjects = ["Python", "Maths", "Physics"]
marks = df[subjects].to_numpy()

print("Subject averages:", np.mean(marks, axis=0))
print("Highest mark:", np.max(marks))
print("Lowest mark:", np.min(marks))
```

## 5. Pandas analysis
Create at least two calculated columns or useful summaries. Filter and sort rows to answer questions about your dataset.


```python
# TODO: Create calculated column(s)
# TODO: Filter records using a condition
# TODO: Sort records and display the top/bottom results
# Create total marks and percentage columns
subjects = ["Python", "Maths", "Physics"]
df["Total"] = df[subjects].sum(axis=1)
df["Percentage"] = df["Total"] / 3

# Find students with below 40 marks in any subject
below_40 = df[(df[subjects] < 40).any(axis=1)]
print("Students scoring below 40 in any subject:")
display(below_40)

# Show the top 10 students by total marks
top_students = df.sort_values("Total", ascending=False).head(10)
print("Top 10 students:")
display(top_students)
```

## 6. Your individual project challenge
**Challenge:** Calculate total and percentage; identify topper and students below 40 in any subject; compare subject averages.

Write your solution below and explain what the result means.


```python
# TODO: Solve your assigned challenge
subjects = ["Python", "Maths", "Physics"]

# Calculate each student's total marks and percentage
df["Total"] = df[subjects].sum(axis=1)
df["Percentage"] = df["Total"] / 3

# Find the student(s) with the highest total
highest_total = df["Total"].max()
topper = df[df["Total"] == highest_total]

print("Topper:")
display(topper[["Student_ID", "Name", "Total", "Percentage"]])

# Find students who scored below 40 in any subject
below_40 = df[(df[subjects] < 40).any(axis=1)]

print("Students scoring below 40 in any subject:")
display(below_40[["Student_ID", "Name"] + subjects])

# Compare the average marks in each subject
print("Average marks by subject:")
print(df[subjects].mean())
```
